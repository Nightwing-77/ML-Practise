```python
%%writefile Matrix_transpose_Tiling.cu

#include <iostream>
#include <cmath>
#define TILE_DIM 32

__global__ voide Matrix_transpose_Tiling(const float *input, float *output, int widht,int height){

  __shared__ float tile[TILE_DIM][TILE_DIM+1];

  int x= blockIdx.x*TILE_DIM+threadIdx.x;
  int y= blockIdx.y*TILE_DIM+threadIdx.y;

  if (x<widht && y<height){
    int input_index=y*widht+x;
    tile[threadIdx.y][threadIdx.x]=input[input_index];
  }
  __syncthreads();


  int x= blockIdx.y*TILE_DIM+threadIdx.y;
  int y= blockIdx.x*TILE_DIM+threadIdx.x;

  if(x<height && y<width){
    int output_index=y*height+x;
    output[output_index]=tile[threadIdx.x][threadIdx.y];
  }

}
```
